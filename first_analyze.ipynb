```python
import pandas as pd
import numpy as np


# 我在仓库被其他人更新的时候，修改了我的本地文件（本行内容和仓库的内容不一致）
# 说明：这是一个sxxxxx文件
''' 这是一个示例数据框 
dawdasdawDjdioasjcxoiad
sdjaiwodjasknmcxoanszo
'''

data = pd.DataFrame({
    "sample_id": [1, 2, 3, 4, 5, 6],
    "group": ["control"]*3 + ["treatment"]*3,
    "activity": [90, 100, 110, 120, np.nan, 140],
})
# data是什么？ 三列六行的数据；

summary1 = data.groupby("group")["activity"].agg(["count", "mean"])  
#这一行在做什么？先对group列根据字段分类,只选中 activity 这一列来操作;
#  再对每一类进行计数和平均值计算(count/mean)；
summary1  #summary1存储了操作后的数据；xiuxiuxiugai 
```
